```python
from statsmodels.stats.power import NormalIndPower
import numpy as np

effect_size = 0.8
alpha = 0.05
power = 0.8

analysis = NormalIndPower()
sample_size = analysis.solve_power(effect_size=effect_size, power=power, alpha=alpha)
print("Gerekli örneklem büyüklüğü:", 6 * np.ceil(sample_size))
```

```python
import pandas as pd
melih = pd.read_csv("tracks_labeled_melih.csv", sep=";")
tarik = pd.read_csv("tracks_labeled_tarik.csv")

print("Melih'in etiketlediği şarkı sayısı:", len(melih))
print("Tarık'ın etiketlediği şarkı sayısı:", len(tarik))
```

```python
from sklearn.metrics import cohen_kappa_score

labels_melih = melih['label_name'].tolist()
labels_tarik = tarik['predicted_label_name'].tolist()


print("Cohen's Kappa:", cohen_kappa_score(labels_melih, labels_tarik))
```

```python
df_combined = melih.copy()
df_combined['tarik_label'] = tarik['predicted_label_name']

# Uyumsuz satırları filtreleyin:
df_mismatch = df_combined[df_combined['label_name'] != df_combined['tarik_label']]

# Tüm uyuşmayan verileri gösterin:
df_mismatch
```
